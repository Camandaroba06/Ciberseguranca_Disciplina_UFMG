```python
def gerar_chave(tamanho: int) -> bytes:
    """Gera uma chave criptograficamente segura com tamanho bytes."""
    chave_gerada_aleatoriamente = secrets.token_bytes(tamanho)
    return chave_gerada_aleatoriamente

def xor_bytes(dados: bytes, chave: bytes) -> bytes:
    """Calcula o XOR entre sequencias de mesmo comprimento."""
    if(len(dados)==len(chave)):
        # resultado_xor = bytearray()
        resultado_xor = []
        for i in range(len(dados)):
            resultado_xor.append(dados[i]^chave[i]) 
        return resultado_xor
    else:
        raise ValueError("Erro! Tamanhos diferentes entre dados e chave!")


def cifrar(mensagem: bytes, chave: bytes) -> bytes:
    """Cifra mensagem usando OTP."""
    if(len(mensagem)==len(chave)):
        cifrado_calculado = xor_bytes(mensagem,chave)
        return cifrado_calculado
    else:
        raise ValueError("Erro! Tamanhos diferentes entre mensagem e chave!")

def decifrar(cifrado: bytes, chave: bytes) -> bytes:
    """Decifra um texto cifrado usando OTP."""
    if(len(cifrado)==len(chave)):
        mensagem_decifrada = xor_bytes(cifrado,chave)
        return mensagem_decifrada
    else:
        raise ValueError("Erro! Tamanhos diferentes entre cifrado e chave!")
```

A diferença entre o número de caracteres de uma string e o seu tamanho em bytes ocorre porque o padrão UTF-8 utiliza um sistema de codificação de tamanho variável. Isso significa que ele não gasta a mesma quantidade de espaço na memória para representar todas as letras do nosso alfabeto ou símbolos digitais.

Caracteres básicos da tabela original ASCII, como letras de "A" a "Z" sem acentuação, números e marcações de pontuação simples, são muito padronizados e ocupam exatamente um byte cada. No entanto, quando utilizamos caracteres mais complexos ou regionais, como letras acentuadas ("á", "ç", "ã"), símbolos matemáticos, emojis ou até alfabetos inteiros de outros idiomas (como o mandarim ou o cirílico), o computador precisa de um intervalo numérico maior para conseguir representar todas essas opções únicas.

Para resolver isso, o UTF-8 passa a alocar de dois a quatro bytes para renderizar um único caractere visual. É por esse motivo que, na prática, uma palavra de quatro letras sem acentos (como "pato") terá exatamente quatro bytes de tamanho, enquanto uma palavra também de quatro letras, mas com acentuação (como "ação"), ocupará seis bytes de espaço na memória, já que o "ç" e o "ã" exigirão dois bytes cada um para serem processados corretamente pelo sistema.
